### Veri Analizi Gözüyle Değişkenler ve Veri Tipleri

```python
# Kirli veriyi temizleyip doğru veri tipine dönüştürme (Type Casting) 
ham_fiyat = " 1250.75 TL "
temiz_fiyat=float(ham_fiyat.replace("TL", "").strip()) 
print(f"Değer: {temiz_fiyat} | Tipi: {type(temiz_fiyat)} | KDV Dahil: {temiz_fiyat * 1.20:.2f}")
```

#### 1.UYGULAMA CEVABI

```python
# Başlangıç Verisi 
urun_kodu = " urn-4092 " 
birim_fiyat_str = "850.50"
siparis_adedi_str = "4" 
indirim_orani_str= "15" 
# 1. Adım: Ürün kodunu temizleme ve standartlaştırma 
temiz_kod = urun_kodu.strip().upper() 
# 2. Adım: Veri tipi dönüşümleri (Type Casting) 
birim_fiyat = float(birim_fiyat_str) 
siparis_adedi = int(siparis_adedi_str) 
indirim_orani= float(indirim_orani_str) 
# 3. Adım: Hesaplama ve formatlı raporlama 
brut_tutar = birim_fiyat * siparis_adedi 
indirimli_toplam = brut_tutar * (1 - indirim_orani / 100) 
print(f"Ürün: {temiz_kod} | Adet: {siparis_adedi} | İndirimli ToplamTutar: {indirimli_toplam:.2f} TL")
```

#### 2.UYGULAMA CEVABI

```python
gunluk_ciro = 18500.0
gunluk_maliyet = 12200.0 
iade_adedi = 14 
toplam_islem_adedi = 120 
sistem_notu = "POS bağlantısında gecikme ve HATA tespit edildi" 
# 1. ve 2. Adım: Oranların hesaplanması 
kar_orani = ((gunluk_ciro - gunluk_maliyet) / gunluk_ciro) * 100 
iade_orani = (iade_adedi / toplam_islem_adedi) * 100 
# 3. ve 4. Adım: Mantıksal bayrak(Flag) değişkenleri 
basarili_gun = (kar_orani > 30) and (iade_orani < 10) 
inceleme_gerekli = (iade_orani >= 10) or ("hata"in sistem_notu.lower()) 
print(f"Kâr Oranı: %{kar_orani:.2f} | İade Oranı: %{iade_orani:.2f}") 
print(f"Başarılı Günmü?:{basarili_gun} | İnceleme Gerekli mi?: {inceleme_gerekli}")
```

### Koşullu İfadeler (if - elif - else): Veri Segmentasyonu

```python
#Müşteri Harcama Tutarına Göre Segmentasyon ve Veri Doğrulama 
harcama = 4800
if harcama < 0: 
    segment = "HATALI VERİ(Negatif Harcama)" 
elif harcama == 0:
     segment = "Pasif Müşteri" 
elif harcama < 2000:
     segment = "Standart Segment" 
elif harcama < 5000: 
    segment = "Gümüş Segment" 
else: 
    segment = "Altın (VIP) Segment" 
print(f"Harcama: {harcama} TL -> AtananSegment: {segment}")
```

### Döngüler (for): Veri Seti Satırları Üzerinde Gezinme

```python
# 6 Günlük Satış Verisi Üzerinde Toplam, Ortalama ve Hedef Kontrolü 
gunluk_satislar = [1200, 1850, 950, 2400, 1600, 3100] 
toplam_satis = 0 
hedefi_asan_gun_sayisi = 0 
for satis in gunluk_satislar: 
    toplam_satis += satis # Kümülatif toplam (Akümülatör) 
    if satis >= 1500: 
        hedefi_asan_gun_sayisi += 1 # Koşullu sayaç (Counter) 
ortalama = toplam_satis / len(gunluk_satislar) 
print(f"Toplam: {toplam_satis} TL | Ortalama: {ortalama:.2f} TL | Hedefi Aşan Gün: {hedefi_asan_gun_sayisi}")
```

#### Uygulama 5 Cevabı

```python
haftalik_ciro = [14500, 19200, 11800, 24600, 17900, 28400, 21000] 
toplam_ciro = 0
yuksek_ciro_gun_sayisi=0 
maksimum_ciro = haftalik_ciro[0] 
en_iyi_gun = 1 
for gun_no, ciro in enumerate(haftalik_ciro, start=1): 
    toplam_ciro+= ciro 
    if ciro >= 20000: 
        yuksek_ciro_gun_sayisi += 1 
    if ciro > maksimum_ciro: 
        maksimum_ciro = ciro 
        en_iyi_gun = gun_no 
ortalama_ciro = toplam_ciro / len(haftalik_ciro)
print(f"Haftalık Toplam: {toplam_ciro} TL | Günlük Ortalama: {ortalama_ciro:.2f} TL") 
print(f"20.000 TL Üzeri Gün Sayısı: {yuksek_ciro_gun_sayisi} gün") 
print(f"Rekor Ciro: {en_iyi_gun}. Gün ({maksimum_ciro} TL)")
```

### While Döngüsü ve Döngü Kontrolü (break & continue)

```python
# Kirli Veri Akışında: Eksik/Negatif Kayıtları Atla (continue), Kritik Hatada Dur (break) 
veri_akisi = [120, 150, -99, 200, None, 180, 99999, 140] 
gecerli_toplam = 0 
gecerli_adet = 0 
for deger in veri_akisi: 
    if deger is None or deger< 0: 
        print(f"Uyarı: Geçersiz kayıt ({deger}) atlandı.") 
        continue # Bu satırı hesaba katma, sıradaki elemana geç! 
    if deger == 99999: 
        print("Kritik Durdurma Kodu (99999) alındı! Veri okuma sonlandırılıyor.")
        break # Döngüyü tamamen bitir! 
    gecerli_toplam += deger 
    gecerli_adet += 1 
print(f"Geçerli Kayıt: {gecerli_adet} | Temiz Ortalama: {gecerli_toplam / gecerli_adet:.2f}")
```

#### Uygulama 6 Cevabı

```python
faturalar = [2400, -150, 3100, 0, 2800, None, 2900, 1500] 
butce_limiti = 10000 
onaylanan_toplam = 0  
onaylanan_fatura_sayisi = 0 
hatali_kayit_sayisi = 0 
for fatura in faturalar: 
    if fatura is  None or fatura <= 0: # 1. Adım: Eksik veya geçersiz verikontrolü 
        hatali_kayit_sayisi += 1 
        continue 
    if onaylanan_toplam+ fatura>butce_limiti: # 2. Adım: Bütçe aşım kontrolü 
        print(f"Uyarı: {fatura} TL tutarındaki fatura bütçeyi aşıyor. Bütçe limiti doldu!") 
        break 
    #3.Adım:Geçerli faturayı onaylama 
    onaylanan_toplam += fatura 
    onaylanan_fatura_sayisi += 1 
    kalan_butce = butce_limiti- onaylanan_toplam 
print(f"Onaylanan: {onaylanan_fatura_sayisi} Adet ({onaylanan_toplam} TL) | Kalan Bütçe: {kalan_butce} TL|Hatalı: {hatali_kayit_sayisi}")
```

#### Uygulama 7 çözümü

```python
ham_kayitlar = ["1200", " 3400 ", "İPTAL", "-500", "850", "5200", "BOS", "2100", "900"]
gecerli_toplam = 0
gecerli_adet = 0 
gecersiz_sayisi = 0 
dusuk_satis, normal_satis, yuksek_satis = 0, 0, 0 
for ham in ham_kayitlar: 
    temiz= ham.strip() 
    if not temiz.isdigit(): 
        gecersiz_sayisi += 1 
        continue 
    tutar = int(temiz) 
    gecerli_toplam += tutar 
    gecerli_adet += 1 
    if tutar < 1000: 
        dusuk_satis += 1 
    elif tutar <= 3000: 
        normal_satis += 1 
    else: 
        yuksek_satis += 1 
ortalama = gecerli_toplam / gecerli_adet 
print(f"Geçerli: {gecerli_adet} | Geçersiz/Elendi: {gecersiz_sayisi} | Toplam: {gecerli_toplam} TL | Ort: {ortalama:.2f} TL") 
print(f"Dağılım -> Düşük: {dusuk_satis} | Normal: {normal_satis} | Yüksek: {yuksek_satis}")
```

#### Uygulama 8 Cevabı

```python
sepetler = [600, 1400, 350, -100, 2200, 900] 
kuponlar = ["INDIRIM10", "VIP20", "YOK", "INDIRIM10", "VIP20", "INDIRIM10"] 
net_gelir, toplam_indirim = 0.0, 0.0 
for i in range(len(sepetler)): 
    tutar, kupon = sepetler[i], kuponlar[i] 
    if tutar<= 0: 
        continue
    indirim = 0.0 
    if kupon == "INDIRIM10" and tutar > 500: 
        indirim = tutar * 0.10 
    elif kupon == "VIP20" and tutar > 1000:
        indirim = tutar * 0.20 
    net_gelir += (tutar - indirim)
    toplam_indirim += indirim 
print(f"Net ToplamGelir: {net_gelir:.2f} TL | Toplam Verilen İndirim: {toplam_indirim:.2f} TL")
```
