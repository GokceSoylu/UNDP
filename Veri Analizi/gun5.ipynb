```python
import pandas as pd

# Veri setinin oluşturulması
siparis = pd.DataFrame({
    "Sehir": ["Karabuk", "Ankara", "Istanbul", "Karabuk", "Ankara", "Karabuk", "Istanbul", "Ankara"],
    "Kategori": ["Elektronik", "Giyim", "Elektronik", "Kitap", "Elektronik", "Giyim", "Kitap", "Giyim"],
    "Tutar": [4500, 850, 6200, 320, 3900, 740, 280, 1100],
    "Puan": [5, 4, 5, 4, 3, 5, 5, 4]
})
en_yuksek_3 = siparis.sort_values(by="Tutar", ascending=False).head(3)
print(en_yuksek_3)
sehir_frekans = siparis["Sehir"].value_counts()
print(sehir_frekans)
kategori_yuzde = siparis["Kategori"].value_counts(normalize=True) * 100
print(kategori_yuzde)
kategori_yuzde = siparis["Kategori"].value_counts()
print(kategori_yuzde)
```
