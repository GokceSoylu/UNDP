## Listeler (List): İndeksleme, Dilimleme (Slicing) ve Metotlar

```python
# Aylık Satış Verisi Üzerinde Dilimleme (Slicing) ve Hazır İstatistikler 
aylik_satislar = [120, 135, 150, 140, 180,210, 195, 230] 
ilk_uc_ay = aylik_satislar[:3] 
# 0, 1 ve 2. indeksler -> [120, 135, 150] 
son_iki_ay = aylik_satislar[-2:] 
# Sondan 2 eleman -> [195, 230] 
ikinci_ceyreklik = aylik_satislar[3:6] 
# 3, 4 ve 5. indeksler -> [140, 180, 210] 
# # Veri Ekleme, Sıralama ve Özet Fonksiyonlar 
aylik_satislar.append(250) 
# Sona yeni ay verisi ekler 
sirali_satis = sorted(aylik_satislar) 
# Orijinal listeyi bozmadan küçükten büyüğe sıralar 
print(f"İlk 3 Ay: {ilk_uc_ay} | Son 2 Ay: {son_iki_ay}") 
print(f"Min: {min(aylik_satislar)} | Max: {max(aylik_satislar)} | Ort: {sum(aylik_satislar)/len(aylik_satislar):.1f}")
```

### Çözüm 1: Çeyreklik Analizi ve Uç Değer Kırpma

```python
tuketim = [420, 390, 350, 310, 280, 260, 275, 980, 300, 340, 380, 110] 
# 1. Adım: İlk Çeyrek (İlk 3 ay) ve Son  Çeyrek (Son 3 ay) dilimleme 
q1_tuketim = tuketim[:3] 
q4_tuketim = tuketim[-3:] 
# 2. Adım: Tüm ayların normal ortalaması 
normal_ort = sum(tuketim) / len(tuketim) 
# 3. Adım: En küçük ve en büyük uç değeri kırparak ortalama (Trimmed Mean) 
sirali_tuketim = sorted(tuketim) 
kirpilmis_liste = sirali_tuketim[1:-1] 
# İlk (min) ve son (max) elemanı dışarıdabırakır 
kirpilmis_ort = sum(kirpilmis_liste) / len(kirpilmis_liste) 
print(f"Q1 Toplam: {sum(q1_tuketim)} MWh ({q1_tuketim})|Q4 Toplam: {sum(q4_tuketim)} MWh ({q4_tuketim})") 
print(f"Normal Ortalama: {normal_ort:.2f} MWh | Kırpılmış Ortalama: {kirpilmis_ort:.2f} MWh")
```

## Demetler (Tuple) ve Kümeler (Set): Tekilleştirme & Kesişim

```python
# 1. Tuple Unpacking (Çoklu Değişken Çözümleme) 
veri_boyutu = (15000, 12) 
# (Satır Sayısı, Sütun Sayısı)
satir_sayisi, sutun_sayisi = veri_boyutu 
# 2. Set ile Tekrarlayan (Duplicate) Müşteri ID'lerini Temizleme 
ham_ziyaretciler = [101, 102, 101, 105, 102, 108, 101] 
tekil_ziyaretciler = set(ham_ziyaretciler) 
# {101, 102, 105,108} # 3. Küme İşlemleri: Kesişim (&), Birleşim (|), Fark (-) 
web_musterileri = {"M1", "M2", "M3", "M4"} 
mobil_musterileri= {"M3", "M4", "M5"} 
her_iki_kanal = web_musterileri & mobil_musterileri 
# Kesişim: {'M3', 'M4'} 
sadece_web = web_musterileri - mobil_musterileri 
# Fark: {'M1', 'M2'} 
print(f"Tekil Ziyaretçi: {len(tekil_ziyaretciler)} | Ortak Kanal:{her_iki_kanal} | Sadece Web: {sadece_web}")
```

### Çözüm 2: Müşteri Kaybı (Churn) ve Tekilleştirme Analizi

```python
ekim_siparis_id = ["U101", "U102", "U101", "U103", "U104", "U105", "U102", "U106"] 
kasim_siparis_id=["U103","U104", "U104", "U107", "U108", "U103", "U109"] 
# 1. Adım: Tekilleştirme ve kopya (duplicate)sayısı
ekim_tekil=set(ekim_siparis_id) 
kasim_tekil = set(kasim_siparis_id) 
ekim_kopya_sayisi = len(ekim_siparis_id)-len(ekim_tekil)
#2,3 ve 4. Adım: Küme kesişim ve fark analizleri 
sadik_musteriler = ekim_tekil & kasim_tekil 
# Her iki ayda da var olanlar 
kayip_musteriler = ekim_tekil - kasim_tekil 
# Ekim'de var, Kasım'da yok (Churn) 
yeni_musteriler=kasim_tekil- ekim_tekil 
# Kasım'da var, Ekim'de yok (Yeni) 
print(f"Ekim Tekil Müşteri: {len(ekim_tekil)} |Ekim Kopya Kayıt:{ekim_kopya_sayisi}") 
print(f"Sadık Müşteriler ({len(sadik_musteriler)}): {sorted(sadik_musteriler)}")
print(f"Kaybedilen/ Churn ({len(kayip_musteriler)}): {sorted(kayip_musteriler)}") 
print(f"Yeni Kazanılan ({len(yeni_musteriler)}): {sorted(yeni_musteriler)}")
```

## Sözlükler (Dictionary): Anahtar-Değer Yapısı ve Frekans Tablosu

```python
# 1. Sözlük Tanımı ve .get() ile Güvenli Okuma (KeyError Hatasını Önler) 
urun_bilgi = {"kod": "URN-10", "fiyat": 450.0, "stok": 25} 
indirim = urun_bilgi.get("indirim_orani", 0.0) 
# Anahtar yoksa hata vermez, 0.0 döner! 
# 2. Kategorik Veri Üzerinde Frekans (Tekrar Sayısı) Hesaplama 
sikayet_kategorileri = ["Kargo", "İade", "Kargo", "Ürün", "Kargo", "İade"] 
frekans_tablosu = {} 
for kategori in sikayet_kategorileri: 
    frekans_tablosu[kategori] = frekans_tablosu.get(kategori, 0) + 1 
    
# 3. .items() ile Anahtar ve Değerleri Birlikte Gezme 
for kat, adet in frekans_tablosu.items(): 
    print(f"Kategori: {kat} -> Şikayet Adedi: {adet}")
```

### Çözüm 3: Şehir Bazlı Sipariş Frekansı ve Ciro Özeti

```python
siparisler = [ ("Karabük", 1200), ("Bartın", 850), ("Karabük", 2300), ("Zonguldak", 1500), ("Karabük", 900), ("Zonguldak", 2100),("Bartın", 1150), ("Kastamonu", 1800) ] 
sehir_adet = {}
sehir_ciro = {} 
# 1. ve 2. Adım: Şehir bazındafrekans ve toplam ciro biriktirme 
for sehir, tutar in siparisler: 
    sehir_adet[sehir] = sehir_adet.get(sehir, 0)+ 1 
    sehir_ciro[sehir] = sehir_ciro.get(sehir, 0) + tutar 
    # 3. Adım: Şehir özet tablosunu raporlama 
for sehir, toplam_tutar in sehir_ciro.items(): 
    adet = sehir_adet[sehir] 
    ort_sepet = toplam_tutar / adet 
    print(f"{sehir:10s} -> Sipariş: {adet} Adet|Toplam: {toplam_tutar} TL | Ort. Sepet: {ort_sepet:.2f} TL")
```

## İç İçe Veri Yapıları: Sözlük Listeleri (Tablo ve JSON Yapısı)

```python
# Tablo Yapısı: Her sözlük bir satırı (gözlemi), her anahtar bir sütunu temsil eder 
personel_tablosu = [ {"isim": "Ayşe", "departman": "Veri Analizi", "maas": 42000, "deneyim": 4}, {"isim": "Burak", "departman": "Yazılım", "maas": 38000, "deneyim": 2}, {"isim": "Ceren", "departman": "Veri Analizi", "maas": 51000, "deneyim": 6} ] 
# Veri Analizi Departmanındaki Personelleri Filtreleme ve Yeni Sütun ("prim") Ekleme 
for satir in personel_tablosu: 
    satir["prim"] = satir["maas"] * 0.10 if satir["deneyim"] >= 3 else 0.0 
    if satir["departman"] == "Veri Analizi": 
        print(f"{satir['isim']}| Maaş:{satir['maas']} TL | Hesaplanan Prim: {satir['prim']:.0f} TL")
```

## Fonksiyonlar (def): Tekrar Eden Analizleri Otomatikleştirme

```python
# Varsayılan Parametreli ve Çoklu Değer Döndüren (Tuple Return) Analiz Fonksiyonu 
def fiyat_temizle_ve_hesapla(ham_metin, kdv_orani=0.20):
     """Kirli fiyat metnini temizler, net ve KDV dahil fiyatı döndürür.""" 
     temiz = ham_metin.replace("TL", "").replace(",", ".").strip() 
     if not temiz: 
        return 0.0, 0.0 
     net_fiyat = float(temiz) 
     kdv_dahil = round(net_fiyat * (1 + kdv_orani), 2) 
     return net_fiyat, kdv_dahil 
# Fonksiyonun Çağrılması (Varsayılan %20 KDV ve Özel %10 KDV ile)
net1, brüt1 = fiyat_temizle_ve_hesapla(" 1500,50 TL ") 
net2, brüt2 = fiyat_temizle_ve_hesapla(" 800 TL ", kdv_orani=0.10) 
print(f"1. Ürün -> Net: {net1} | KDV Dahil: {brüt1} || 2. Ürün -> Net:{net2} | KDV Dahil: {brüt2}")
```

### Çözüm 4: Otomatik Veri Özeti Fonksiyonu Yazımı

```python
def veri_ozeti_cikar(veri_listesi): 
    # 1. Adım: Eksik ve geçersiz verileri filtreleme 
    temiz_veri = [x for x in veri_listesi if x is not None and x > 0] 
    if len(temiz_veri) == 0: 
        return {"hata": "Geçerli veri bulunamadı"}
    # 2. Adım: Özet istatistikleri sözlük olarak döndürme 
    return { "gecerli_adet": len(temiz_veri), "minimum": min(temiz_veri), "maksimum": max(temiz_veri), "ortalama": round(sum(temiz_veri) / len(temiz_veri), 2) } 
magaza_a_satis = [1200, 1500, -200, None, 1800, 2500, 0] 
magaza_b_satis = [3200, 4100, 2900, None, 5000] 
print("Mağaza A Özeti:", veri_ozeti_cikar(magaza_a_satis)) 
print("Mağaza B Özeti:", veri_ozeti_cikar(magaza_b_satis))
```

## Lambda (Anonim) Fonksiyonlar: Tek Satırlık Hızlı İşlemler

```python
# 1. Temel Lambda Tanımı: lambda parametre: donus_degeri 
kdv_ekle = lambda fiyat: round(fiyat * 1.20, 2) 
print(f"100 TL KDV Dahil: {kdv_ekle(100)} TL") 
# 2. En Sık Kullanım: Sözlük Listesini (Tabloyu) İstediğimiz SütunaGöre Sıralama! 
adaylar = [ {"isim": "Zeynep", "puan": 82}, {"isim": "Ahmet", "puan": 95}, {"isim": "Mehmet", "puan": 74}] 
#Puan sütununa göre büyükten küçüğe (reverse=True) sıralama 
sirali_adaylar = sorted(adaylar, key=lambda x: x["puan"], reverse=True) 
print("Birinci Aday:", sirali_adaylar[0]) 
# {'isim': 'Ahmet', 'puan': 95}
```

### Çözüm 5: Lambda ile Müşteri Skorlama ve Sıralama

```python
musteriler = [ {"id": "M101", "harcama": 120000, "gecikme": 2}, {"id": "M102", "harcama": 85000, "gecikme":0},{"id": "M103", "harcama": 150000, "gecikme": 5}, {"id": "M104", "harcama": 95000, "gecikme": 1} ] 
# 1. Adım: İki parametreli lambda fonksiyonu ile skor formülü 
skor_hesapla = lambda harcama, gecikme: (harcama / 1000) - (gecikme*15)
# 2. Adım: Her müşteriye "skor" sütunu ekleme 
for m in musteriler: m["skor"] = skor_hesapla(m["harcama"], m["gecikme"]) 
#3. Adım: Skor sütununa göre büyükten küçüğe sıralama 
sirali_musteriler = sorted(musteriler, key=lambda x: x["skor"], reverse=True) 
for sira, m in enumerate(sirali_musteriler, start=1): 
    print(f"{sira}. Sıra -> ID: {m['id']} | Harcama: {m['harcama']} TL | Gecikme: {m['gecikme']} | Skor: {m['skor']}")
```

### Çözüm 6: Uçtan Uca Veri Yapılandırma Hattı (Pipeline)

```python
ham_satirlar = [ "Karabük;Elektronik;4500", "Bartın;Giyim;1200", "Karabük;Elektronik;4500", "Zonguldak;Ev;HATALI", "Karabük;Giyim;1800", "Bartın;Elektronik;3200", "Zonguldak;Elektronik;5100" ] 
# 1. Adım: Tekrarlayan (Duplicate) satırları set() ile temizleme 
tekil_satirlar = list(set(ham_satirlar)) 
# 2. Adım: Satır ayrıştırıcı ve doğrulayıcı fonksiyon 
def satir_ayristir(satir): 
    sehir, kategori, tutar_str = satir.split(";") 
    if not tutar_str.strip().isdigit(): 
        return None 
    return {"sehir": sehir.strip(), "kategori": kategori.strip(), "tutar": int(tutar_str.strip())} 
# List Comprehension ile temiz sözlük tablosu üretme 
temiz_tablo = [satir_ayristir(s) 
for s in tekil_satirlar if satir_ayristir(s) is not None] 
# 3. Adım: Kategori bazlı toplam ciro analizi 
kategori_ciro = {} 
for kayit in temiz_tablo: 
    kat = kayit["kategori"] 
    kategori_ciro[kat] = kategori_ciro.get(kat, 0) + kayit["tutar"] 
print(f"Ham Satır: {len(ham_satirlar)} | Tekil ve Geçerli Kayıt: {len(temiz_tablo)}") 
print(f"Kategori Bazlı Ciro Özeti: {kategori_ciro}")
```
