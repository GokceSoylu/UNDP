## 1. BÖLÜM • MATPLOTLİB'E GİRİŞ

## Her Grafikte Kullanacağınız 5 Satırlık Temel Şablon


```python
# Örnek Veri Seti
kategoriler = ["Market", "Kira", "Faturalar", "Ulaşım", "Eğlence"]
degerler = [4500, 12000, 2100, 1500, 1800]
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4)) 
# 1. Grafik boyutunu belirle (Genişlik, Yükseklik) 
plt.bar(kategoriler, degerler) 
# 2. Grafik türünü çiz (bar, plot veya hist) 
plt.title("Grafik Başlığı") 
# 3. Anlaşılır bir başlık ekle 
plt.xlabel("Yatay Eksen (X)") 
# 4. X ve Y eksen adlarını yaz 
plt.ylabel("Dikey Eksen (Y)") 
plt.show()
# 5. Grafiği ekranda temiz biçimde göster
```

# Ek örnek Seviye 2 

## ADIM A

```python
 # Kütüphanelerin Yüklenmesi ve Veri Setinin Oluşturulması
import pandas as pd
import matplotlib.pyplot as plt

# A.1. Veri: Sipariş kayıtları (Pandas DataFrame)
veri = {
    "Sehir": [
        "İstanbul", "Ankara", "İzmir", "Karabük",
        "İzmir", "Karabük", "İstanbul", "Ankara",
        "İstanbul", "Ankara", "İzmir", "Karabük",
        "İstanbul", "Ankara", "İzmir", "Karabük",
        "İstanbul", "Ankara", "İzmir", "Karabük",
        "İstanbul", "Ankara", "İzmir", "Karabük"
    ],
    "Kategori": [
        "Burger", "Pizza", "Kebap", "Burger", "Kahve", "Tatlı",
        "Pizza", "Kebap", "Burger", "Kebap", "Kahve", "Pizza",
        "Tatlı", "Burger", "Kebap", "Kahve", "Pizza", "Kebap",
        "Burger", "Tatlı", "Kebap", "Pizza", "Burger", "Kebap"
    ],
    "Siparis_Tutari": [
        340, 420, 510, 280, 190, 260, 390, 460,
        360, 540, 210, 350, 290, 310, 490, 180,
        450, 520, 330, 240, 580, 410, 350, 470
    ],
    "Teslimat_Suresi": [
        48, 32, 29, 25, 52, 35, 44, 22, 55, 38,
        27, 30, 46, 33, 41, 24, 58, 36, 31, 28,
        49, 42, 34, 26
    ],
    "Puan": [
        3.2, 4.5, 4.7, 4.8, 2.9, 4.4, 3.5, 4.9,
        2.5, 4.1, 4.6, 4.5, 3.1, 4.3, 3.8, 4.7,
        2.4, 4.2, 4.4, 4.6, 3.0, 3.6, 4.3, 4.8
    ]
}
df = pd.DataFrame(veri)
```

### A.2.

```python

# Veri: 2025 ve 2026 Aylık Sipariş Sayıları (Bin Adet)
aylar = ["Oca", "Şub", "Mar", "Nis", "May", "Haz"]
siparis_2025 = [42, 45, 48, 51, 55, 60]
siparis_2026 = [48, 53, 50, 62, 68, 79]

print("Veri seti başarıyla yüklendi. Toplam kayıt sayısı:", len(df))
```

## ADIM B 

```python
# Grafik 1 - Kategorilere Göre Toplam Ciro (Yatay Sütun Grafiği)
# Kategorilere göre toplam ciroyu hesaplayıp küçükten büyüğe sıralıyoruz
kategori_ciro = df.groupby("Kategori")["Siparis_Tutari"].sum().sort_values()

plt.figure(figsize=(8, 4))
plt.barh(kategori_ciro.index, kategori_ciro.values, color="#1d4ed8")
plt.title("Kategorilere Göre Toplam Sipariş Cirosu")
plt.xlabel("Toplam Ciro (TL)")
plt.ylabel("Kategori")
plt.grid(axis="x", linestyle="--", alpha=0.4)  # Okunabilirliği artırmak için dikey kılavuz çizgisi
plt.tight_layout()
plt.show()
```

## ADIM C 

```python

# Grafik 2 - 2025 ve 2026 Aylık Sipariş Karşılaştırması (Çift Çizgi)
plt.figure(figsize=(8, 4))
plt.plot(aylar, siparis_2025, label="2025", color="#64748b", marker="o", linewidth=2)
plt.plot(aylar, siparis_2026, label="2026", color="#0f766e", marker="o", linewidth=2)
plt.title("2025 ve 2026 Aylık Sipariş Sayısı Karşılaştırması")
plt.xlabel("Aylar")
plt.ylabel("Sipariş Sayısı (Bin Adet)")
plt.legend()  # 2025 ve 2026 etiketlerini kutu içinde gösterir
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## ADIM D

```python
 # Grafik 3 - Teslimat Süresi Dağılımı ve Ortalama Çizgisi (Histogram)
ort = df["Teslimat_Suresi"].mean()  # Ortalama teslimat süresi (36.9 dk)

plt.figure(figsize=(8, 4))
plt.hist(df["Teslimat_Suresi"], bins=7, color="#d97706", edgecolor="white")
plt.axvline(ort, color="red", linestyle="--", linewidth=2, label=f"Ortalama: {ort:.1f} dk")
plt.title("Sipariş Teslimat Süreleri Dağılımı")
plt.xlabel("Teslimat Süresi (Dakika)")
plt.ylabel("Sipariş Sayısı")
plt.legend()
plt.tight_layout()
plt.show()
```

## ADIM E

```python
# Grafik 4 - 40 Dakikayı Aşan Gecikmiş Siparişler (Şehir Bazlı)
# Teslimat süresi 40 dakikadan uzun olan siparişleri filtreliyoruz
geciken = df[df["Teslimat_Suresi"] > 40]
geciken_sehir = geciken["Sehir"].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(geciken_sehir.index, geciken_sehir.values, color="#dc2626")
plt.title("40 Dakikadan Uzun Süren Gecikmiş Siparişlerin Şehirlere Göre Dağılımı")
plt.xlabel("Şehir")
plt.ylabel("Geciken Sipariş Sayısı")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## ADIM F 

```python

# BONUS GÖREV - Şehirlere Göre Ortalama Müşteri Puanı (Koşullu Renk)
sehir_puan = df.groupby("Sehir")["Puan"].mean().sort_values()

# Ortalama puanı 3.5'in altında kalan şehirler kırmızı, diğerleri yeşil olur
renkler = ["#dc2626" if puan < 3.5 else "#16a34a" for puan in sehir_puan.values]

plt.figure(figsize=(8, 4))
plt.bar(sehir_puan.index, sehir_puan.values, color=renkler)
plt.axhline(3.5, color="#334155", linestyle="--", linewidth=1.5, label="Kritik Eşik (3.5 Puan)")
plt.title("Şehirlere Göre Ortalama Müşteri Memnuniyet Puanı")
plt.xlabel("Şehir")
plt.ylabel("Ortalama Puan (1 - 5)")
plt.ylim(0, 5.5)  # Puan eksenini 0-5.5 arası sabitleyerek kıyaslamayı netleştiriyoruz
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Basit Veri Seti (Bir kafenin 5 günlük içecek satışları)
gunler = ["Pzt", "Sal", "Çar", "Per", "Cum"]
kahve = [30, 35, 40, 38, 50]
cay = [45, 42, 48, 46, 55]
limonata = [15, 18, 22, 25, 35]

# 2. Yeni boş tuval açıyoruz
plt.figure(figsize=(8, 4))

# 3. ÜÇ ÇİZGİLİ YAPI: plt.show() demeden art arda 3 kez plt.plot() yazıyoruz
plt.plot(gunler, kahve, label="Kahve", color="#78350f", marker="o", linewidth=2)
plt.plot(gunler, cay, label="Çay", color="#dc2626", marker="s", linewidth=2)
plt.plot(gunler, limonata, label="Limonata", color="#eab308", marker="^", linewidth=2)

# 4. Başlık ve eksen isimleri
plt.title("Hafta İçi İçecek Satışları Karşılaştırması")
plt.xlabel("Günler")
plt.ylabel("Satılan Adet")

# 5. Açıklama kutusu, ızgara ve kenar boşluğu ayarı
plt.legend()                                # Yukarıdaki 3 label'ı kutu içinde gösterir
plt.grid(True, linestyle="--", alpha=0.5)   # Arka plana kesikli ızgara ekler
plt.tight_layout()                          # Yazıların taşmasını/çakışmasını engeller

# 6. Grafiği ekrana basıp tuvali kapatıyoruz
plt.show()
```
