# Atolye1

```python
import pandas as pd
import numpy as np

# Veri setini oluşturma
musteri = pd.DataFrame({
    "Ad": ["Ali", "Ayse", "Mehmet", "Ayse", "Zeynep", "Can"],
    "Sehir": ["Karabuk", "Ankara", np.nan, "Ankara", "Karabuk", "Istanbul"],
    "Yas": [22, 28, 35, 28, np.nan, 24],
    "Harcama": [450, 800, 620, 800, 510, np.nan]
})
```

```python
# 1. Tespit: Her sütundaki eksik değer sayısı
print("--- 1. Eksik Değer Sayıları ---")
print(musteri.isnull().sum())
```

```python
# 2. Genel Silme: Orijinal tabloyu bozmadan eksik değer içeren satırları silme
genel_temiz = musteri.dropna()
print("\n--- 2. Genel Silme Sonucu ---")
print(genel_temiz)
print("Kalan satır sayısı:", len(genel_temiz))  # 3 satır kalır
```

```python
# 3. Hedefli Silme: Sadece Harcama sütunu boş olan satırı silme
hedefli_temiz = musteri.dropna(subset=["Harcama"])
print("\n--- 3. Hedefli Silme (Harcama) Sonucu ---")
print(hedefli_temiz)
```

# fillna

```python
# 1. Sayısal sütunu o sütunun ORTALAMASI ile doldurma 
ort_yas = musteri["Yas"].mean()
musteri["Yas"] = musteri["Yas"].fillna(ort_yas) 
# 2. Sayısal sütunu MEDYAN (ortanca) ile doldurma 
musteri["Harcama"] = musteri["Harcama"].fillna(musteri["Harcama"].median()) 
# 3. Kategorik (metin) sütunu EN SIK GEÇEN değer (Mod) ile doldurma 
en_sik_sehir = musteri["Sehir"].mode()[0] 
musteri["Sehir"] = musteri["Sehir"].fillna(en_sik_sehir)
```

# bölüm 3

```python
# 1. Tabloda kaç adet birebir tekrarlayan satır var? 
musteri.duplicated().sum() 
# 2. Tekrarlayan satırları ekranda görmek için: 
musteri[musteri.duplicated()] 
# 3. Tekrarlayan satırları sil (ilk kaydı korur, kopyaları atar) 
musteri= musteri.drop_duplicates() 
# 4. Sadece belirli bir sütuna göre (örn. Ad) tekrarı engellemek için: 
musteri = musteri.drop_duplicates(subset=["Ad"])
```

# Atolye 2

```python
import pandas as pd
import numpy as np

# Orijinal kirli veri setini oluşturma
musteri = pd.DataFrame({
    "Ad": ["Ali", "Ayse", "Mehmet", "Ayse", "Zeynep", "Can"],
    "Sehir": ["Karabuk", "Ankara", np.nan, "Ankara", "Karabuk", "Istanbul"],
    "Yas": [22, 28, 35, 28, np.nan, 24],
    "Harcama": [450, 800, 620, 800, 510, np.nan]
})
```

```python
# 1. Tekrarlayan kayıt sayısını kontrol etme ve kopya satırı silme
print("Tekrarlayan kayıt sayısı:", musteri.duplicated().sum())
musteri = musteri.drop_duplicates()
```

```python
# 2. Yas sütunundaki eksik değeri ortalama ile doldurma
musteri["Yas"] = musteri["Yas"].fillna(musteri["Yas"].mean())
```

```python
# 3. Harcama sütununu medyan ile, Sehir sütununu "Bilinmiyor" ile doldurma
musteri["Harcama"] = musteri["Harcama"].fillna(musteri["Harcama"].median())
musteri["Sehir"] = musteri["Sehir"].fillna("Bilinmiyor")
```

```python
# 4. Hiç eksik ve tekrar kalmadığını doğrulama
print("\n--- Eksik Değer Kontrolü ---")
print(musteri.isnull().sum())
print("\nKalan tekrar sayısı:", musteri.duplicated().sum())
```

```python
print("\n--- Temizlenmiş Tablo ---")
print(musteri)
```
